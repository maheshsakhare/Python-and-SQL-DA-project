### Read data from the file and handle null values- replace 'Not Available' and 'unknown' values by nan.

```python
df = pd.read_csv('D:/Python_new_Journey_Pong_game/Projects on Pandas/order_sales_analysis/orders.csv',na_values = ['Not Available','unknown'])
print("# Replaced 'Not Available' and 'unknown' values by nan")
df['Ship Mode'].unique()
```

### Rename column names and  ..make them lower case and replace space with underscore

```python
df.columns
```

```python
# Two method are there - 
# df.rename(columns={'Order Id':'order_id','City':'city'}) - but this way we have to enter all the columns.
# Convert all columns names into string.
df.columns = df.columns.str.lower()
```

```python
#Now we will replace spaces from string into underscore.
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
#Derive new columns discount,sale price and profit
df['discount'] = df['list_price']*df['discount_percent']*0.01
```

```python
df['sale_price'] = df['list_price']-df['discount']
```

```python
df['profit'] = df['sale_price']-df['cost_price']
```

```python
df.dtypes
```

```python
#Convert order date from object data type to detetime(use to_datetime method, user % as it is four letters)
df['order_date'] = pd.to_datetime(df['order_date'],format="%Y-%m-%d")
```

```python
df.dtypes
```

```python
#drop cost price list price and discount percent columns('use inplace method to make permanant changes')
df.drop(columns=['list_price','discount_percent','cost_price'],inplace = True)
```
